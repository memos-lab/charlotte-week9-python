## Part A · Decisions with conditions (0–35 minutes)

### Concept introduction: Why programs need decisions

Up to this point, you have written statements that execute in order. Real operational datasets require a program to **choose an action based on information**. For example, a training program might flag a high pressure reading for review, ignore an inactive sensor, or treat a missing measurement as invalid. This is called **conditional execution** or **branching**.

A **condition** is an expression with a Boolean result: `True` or `False`. Python evaluates the condition and runs only the appropriate indented block. A single `if` tests one condition; `elif` means “otherwise, test this different condition”; `else` handles everything not covered earlier.

```python
if pressure_psi >= 3500:
    print("High pressure")
elif pressure_psi >= 2500:
    print("Normal pressure")
else:
    print("Low pressure")
```

**Read it in everyday language:** “If pressure is at least 3500, show high. Otherwise, if it is at least 2500, show normal. Otherwise, show low.” Only **one branch** runs in this `if`/`elif`/`else` chain. Python checks conditions **top to bottom**, so the order matters. The `:` starts a block, and its **indentation** (usually four spaces) determines which statements belong to that branch.

**Key vocabulary:** *condition* (a true/false test), *branch* (code selected by that test), *Boolean* (`True`/`False`), *indentation* (block structure), *threshold* (a decision boundary).

**Worked thought experiment:** At 3500 psi, the first branch runs. At 3000 psi, the second runs. At 2000 psi, the final branch runs. These are fictional **classroom categories, not real equipment operating limits**.

**Common beginner mistakes:** using `=` instead of `==` to compare values; forgetting the colon; inconsistent indentation; placing a broad condition (`>= 2500`) before a narrower one (`>= 3500`); interpreting an `if` result as a sensor safety certification.

**Before running the next cell:** Predict the category for `3100`, then change the value to test your prediction.


````python
pressure_psi = 3100
if pressure_psi >= 3500:
    print("High-pressure training category")
elif pressure_psi >= 2500:
    print("Normal-pressure training category")
else:
    print("Low-pressure training category")
````

### Concept introduction: Comparisons and Boolean logic

A comparison asks a question about two values. `>` and `<` mean greater than and less than; `>=` and `<=` include the boundary; `==` means equal; `!=` means not equal. Each comparison returns a Boolean. **Remember:** `pressure = 2500` stores a value; `pressure == 2500` asks whether it equals 2500.

A realistic rule may require more than one test. Use `and` when **both** conditions must be true (sensor active **and** pressure high), `or` when **at least one** must be true (missing **or** negative), and `not` to reverse a Boolean. Python also supports chained comparisons such as `2500 <= pressure < 3500`.

```python
is_active = status == "active"
is_in_band = 2500 <= pressure_psi < 3500
review = is_active and not is_in_band
```

**Think before you run:** If `status` is `"inactive"`, `review` is false even for a high reading, because both sides of `and` must pass. Conditions should reflect the question the business rule is trying to answer.

**Common mistakes:** writing `status = "active"` inside a condition, writing `pressure >= 2500 or <= 3500` instead of complete comparisons, and confusing `and` with `or`. Use parentheses to make longer expressions easier to read.


````python
pressure_psi = 1800
status = "active"
print("At least 2500 psi:", pressure_psi >= 2500)
print("Active and in normal training band:", status == "active" and 2500 <= pressure_psi < 3500)
print("Not inactive:", not (status == "inactive"))
````

### Concept introduction: Missing and invalid values

A real sensor may be offline, so some measurements are **missing**, not zero. In Python, `None` represents “no value.” It differs from `0` (an actual number) and from an empty string. Attempting `None >= 2500` raises a `TypeError`, because Python cannot compare `None` with an integer that way.

Before classifying a numeric measurement, **validate** it. In this classroom dataset, a negative pressure is considered invalid, so the rule is `value is None or value < 0`. Python evaluates `or` from left to right and **short-circuits**: if the first test is true, it does not evaluate the second. That avoids the invalid numeric comparison.

```python
if value is None or value < 0:
    print("Invalid reading")
else:
    print("Reading can be classified")
```

**Why this matters:** invalid records must not distort summaries or trigger misleading categories. Later in the course, students will build more robust data-quality validation, including handling text and malformed input.

**Common mistakes:** treating `None` as `0`, applying a threshold before checking validity, or using `value == None` rather than the idiomatic `value is None`.


````python
trial_pressures = [3100, None, -8, 1800]
for value in trial_pressures:
    if value is None or value < 0:
        print(value, "→ invalid")
    elif value >= 3500:
        print(value, "→ high")
    elif value >= 2500:
        print(value, "→ normal")
    else:
        print(value, "→ low")
````

## Part B · Repetition with loops (35–75 minutes)

### Concept introduction: Why loops are essential for data analysis

You could write `print(well_1)`, `print(well_2)`, and `print(well_3)`, but that approach becomes impossible when a dataset has thousands of observations. A **loop** lets Python repeat the same operation for every item in a collection. The code stays short even when the data grows.

A `for` loop takes one item at a time from an iterable (such as a list), assigns it to a temporary **loop variable**, and executes the indented body. Each pass through the body is an **iteration**.

```python
well_ids = ["DEMO-01", "DEMO-02", "DEMO-03"]
for well_id in well_ids:
    print("Inspect:", well_id)
```

The first iteration uses `"DEMO-01"`, the second uses `"DEMO-02"`, and so on. Python moves through the list automatically; you do not need to manually track the list index in this example.

**Connection to Class 2:** lists hold multiple readings; dictionaries hold named fields (`record["pressure_psi"]`); loops let you apply the same logic to every record. A common pattern in telemetry analysis is “for each record, check validity, then update a running summary.”

**Key vocabulary:** *iterable*, *iteration*, *loop variable*, *loop body*, *accumulator* (a value built up over repetitions), *filtering* (keeping only records matching a rule).

**Common mistakes:** forgetting the colon after `for`, indenting only the first statement in the loop, accidentally overwriting the input list, or assuming the loop variable refers to the entire list.

**Prediction:** How many lines will the next cell print before the well count?


````python
well_ids = ["DEMO-01", "DEMO-02", "DEMO-03"]
for well in well_ids:
    print("Inspect:", well)
print("Well count:", len(well_ids))
````

### Concept introduction: `range()`, counters and running totals

Use `range()` when you want a known number of repetitions. `range(1, 5)` produces `1, 2, 3, 4`: the **start is included** and the **stop is excluded**. You can also write `range(3)` to obtain `0, 1, 2`.

A **running total** or accumulator begins before the loop, then changes during each iteration. For example, start with `total = 0`, add each valid pressure, and divide by the **number of valid pressures** to calculate an average. The count matters: if you skip invalid records, `len(all_readings)` may not equal the number actually included in the total.

```python
total = 0
count = 0
for pressure in [2200, 2450, 2600]:
    total += pressure
    count += 1
average = total / count
```

**Why this matters in oil and gas:** reports often summarize repeated sensor measurements per well. An accumulator is a building block for daily counts, total production, and mean readings.

**Common mistakes:** resetting `total = 0` *inside* the loop, dividing by zero when no valid samples exist, or assuming `range(1, 5)` includes 5.


````python
for attempt in range(1, 4):
    print(f"Calibration trial {attempt}")

pressures = [2200, 2450, 2600,2800]
total = 0
for pressure in pressures:
    total += pressure
print("Average psi:", total / len(pressures))
````

### Concept introduction: `continue`, `break`, and data filtering

Not all records should contribute to the same calculation. `continue` means **skip the remainder of the current iteration** and go directly to the next item. It is useful when a telemetry record is missing, negative, or from an inactive sensor. `break` means **stop the loop entirely**, which is useful for searches where the first match is enough.

```python
for pressure in [2500, None, 2700]:
    if pressure is None:
        continue
    print(pressure)
```

This prints `2500` and `2700`, not `None`. By contrast, `break` would stop processing any later records.

**Important distinction:** to compute averages for **all** wells, use `continue` to ignore bad records but keep processing. Use `break` only if there is a reason to stop searching the entire collection.

**Common mistakes:** placing `continue` too late (after invalid data was already added), using `break` when the intent was merely to skip one record, or calculating averages using the total number of rows rather than the count of accepted rows.


````python
demo_readings = [
    {"well_id": "DEMO-01", "pressure_psi": 2550, "status": "active"},
    {"well_id": "DEMO-02", "pressure_psi": None, "status": "active"},
    {"well_id": "DEMO-03", "pressure_psi": 2900, "status": "inactive"},
    {"well_id": "DEMO-01", "pressure_psi": 2700, "status": "active"},
    {"well_id": "DEMO-04", "pressure_psi": 3050, "status": "active"},
]
accepted = []
skipped = 0
for record in demo_readings:
    pressure = record["pressure_psi"]
    if record["status"] != "active" or pressure is None or pressure < 0:
        skipped += 1
        continue
    accepted.append(pressure)
print("Accepted readings:", accepted)
print("Skipped:", skipped)
print("Average accepted psi:", sum(accepted) / len(accepted))
````

````python
# Demonstration only: break ends the search once the first high sample is found.
for pressure in [2800, 3000, 3650, 4100]:
    if pressure >= 3500:
        print("First high sample:", pressure)
        break
````

## Part C · Reusable functions (75–115 minutes)

### Concept introduction: From repeated code to reusable functions

When the same conversion or rule appears many times, copying the formula creates opportunities for inconsistent updates and errors. A **function** is a named, reusable block of code that performs one clear task. Define it once with `def`, then **call** it with different inputs whenever needed.

```python
def fahrenheit_to_celsius(temperature_f):
    """Convert Fahrenheit temperature to Celsius."""
    return (temperature_f - 32) * 5 / 9

result = fahrenheit_to_celsius(212)
```

Here, `fahrenheit_to_celsius` is the function **name**, `temperature_f` is a **parameter** (the input placeholder), `212` is an **argument** (the actual input passed in), and `return` sends the computed **result** back to the caller. `result` becomes `100.0`.

**Why return instead of only print?** `print()` displays a value for a person; `return` gives a value back to the program so another calculation can use it. A function without an explicit `return` normally returns `None`.

**Design principle:** a good beginner function has one job, a descriptive name, predictable inputs and outputs, and a simple docstring describing its behavior. Keep domain units in names (`psi`, `kpa`, `temperature_f`) to reduce mistakes.

**Common mistakes:** forgetting parentheses when calling the function, confusing parameters with arguments, putting `print()` where a result must be returned, and expecting variables created inside a function to be automatically accessible outside it (scope).

**Predict first:** If you call `fahrenheit_to_celsius(32)`, what should the function return? Why?


````python
def fahrenheit_to_celsius(temperature_f):
    """Convert a temperature in Fahrenheit to Celsius."""
    return (temperature_f - 32) * 5 / 9

print(fahrenheit_to_celsius(212))
converted = fahrenheit_to_celsius(95)
print(f"95 F = {converted:.1f} C")
````

### Concept introduction: Parameters, default values and return statements

Functions become flexible through **parameters**. A required parameter must be supplied when the function is called. A **default parameter** provides a value when the caller does not specify one: `def describe_well(well_id, status="active")`. Calling `describe_well("DEMO-01")` uses the default; calling `describe_well("DEMO-01", "inactive")` overrides it.

A function may take multiple inputs and return one result (or a tuple/dictionary of results), but keep this week's functions small. The `return` statement also **ends function execution** immediately. This makes it useful for handling invalid values at the top of a function.

**Vocabulary:** *signature* (function name and parameters), *default argument*, *call*, *return value*, *scope*, *docstring* (documentation inside triple quotes).

**Common mistakes:** assuming a default is required, forgetting to pass a required argument, using `return` outside a function, and mistaking code after a `return` for code that will still execute.


````python
def describe_well(well_id, status="active"):
    """Make a short, readable description of a well."""
    return f"{well_id} is {status}"

print(describe_well("DEMO-01"))
print(describe_well("DEMO-02", "inactive"))
````

### Concept introduction: Reusable engineering conversions

A conversion function translates a known formula into code. For pressure, the teaching conversion is **1 psi = 6.8947572932 kPa**. Thus `kpa = psi * 6.8947572932`. For temperature, **Celsius = (Fahrenheit − 32) × 5 / 9**.

Real engineering work also requires awareness of units, missing values, and input constraints. The demonstration pressure function returns `None` for missing or negative inputs. This is a **simple class rule**, not a complete industrial-quality validation specification.

```python
def psi_to_kpa(pressure_psi):
    if pressure_psi is None or pressure_psi < 0:
        return None
    return pressure_psi * 6.8947572932
```

**Check your units:** `2500 psi` becomes about `17236.89 kPa`. Do not round intermediate numbers; format the final display only. In a later class, students will also test their functions systematically.

**Common mistakes:** multiplying by the wrong conversion factor, losing the unit in a variable name, applying formulas to missing input, and rounding too early.


````python
def psi_to_kpa(pressure_psi):
    """Convert a nonnegative pressure from psi to kPa; otherwise return None."""
    if pressure_psi is None or pressure_psi < 0:
        return None
    return pressure_psi * 6.8947572932

print(f"2500 psi = {psi_to_kpa(2500):.2f} kPa")
print("Missing pressure returns:", psi_to_kpa(None))
````

## Part D · Combine collections, decisions, loops and functions (115–150 minutes)

### Concept introduction: Building a small, reusable data workflow

Real analytics programs combine ideas rather than using one idea in isolation. **Collections** store many observations; **loops** visit them; **conditions** decide which observations are valid; **functions** turn repeatable logic into clean operations; and **dictionaries** store a running result for each well.

Imagine data with multiple rows for `TRAIN-A`. If you simply write `pressure_by_well[well_id] = pressure`, a later row **replaces** the earlier one. To calculate an average correctly, store both a **total** and a **count** per well, then compute `total / count`. Skip invalid or inactive records **before** updating either number.

The pattern is:

1. Start with an empty `totals` dictionary and an empty `counts` dictionary.
2. For each record, extract `well_id`, `pressure_psi`, and `status`.
3. Use conditions to skip inactive, missing, or invalid records.
4. Add each accepted pressure to that well's total and increase its count.
5. Calculate a mean for every well with at least one valid observation.
6. Pass the mean to a conversion function if a kPa result is needed.

**Worked mental example:** If one well has valid readings of 2300 and 2500 psi, its total is 4800 and its count is 2, so its mean is 2400 psi. A third valid reading of 2600 changes the mean to `(2300 + 2500 + 2600) / 3 = 2466.67 psi`.

**Why this matters:** this miniature workflow has the same basic logic as much larger telemetry processing pipelines, without needing Pandas or NumPy yet.

**Common mistakes:** overwriting a prior reading, counting rejected records, dividing by zero, or assuming every well has equal numbers of observations.

**Teaching note:** The demonstration data below are different from the formal lab dataset. You will still need to implement your independent lab in a `.py` script.


````python
sample_records = [
    {"well_id": "TRAIN-A", "pressure_psi": 2300, "status": "active"},
    {"well_id": "TRAIN-B", "pressure_psi": 3100, "status": "active"},
    {"well_id": "TRAIN-A", "pressure_psi": 2500, "status": "active"},
    {"well_id": "TRAIN-B", "pressure_psi": 3330, "status": "active"},
    {"well_id": "TRAIN-C", "pressure_psi": None, "status": "active"},
    {'well_id': 'TRAIN_A',  'pressure_psi': 2600, 'status': 'active'},
]
totals = {}
counts = {}
for row in sample_records:
    pressure = row["pressure_psi"]
    if row["status"] != "active" or pressure is None or pressure < 0:
        continue
    well_id = row["well_id"]
    totals[well_id] = totals.get(well_id, 0) + pressure
    counts[well_id] = counts.get(well_id, 0) + 1
for well_id in sorted(totals):
    avg = totals[well_id] / counts[well_id]
    print(f"{well_id}: {avg:.2f} psi from {counts[well_id]} valid active readings")
````
